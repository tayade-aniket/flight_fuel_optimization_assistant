# 11 — Operational What-If Scenario Simulation

### 1. Objective
Evaluate candidate operational adjustments (cruise flight level, cruising speed, and routing distance) on fuel, time, and CO2.

### 2. Aviation Context
Flight dispatchers routinely evaluate whether climbing to a higher flight level or adjusting Mach saves enough fuel to justify minor schedule adjustments.

### 3. Data
Trained fuel model and baseline operational parameters.

### 4. Method
Kinematic feature perturbation and multi-metric delta calculation (fuel kg, time min, CO2 kg, fuel cost USD).


```
import pandas as pd
import joblib
from src.optimization.scenario_simulator import WhatIfSimulator
from src.features.trajectory_features import FeatureExtractor

model = joblib.load("models/best_fuel_model.joblib")
feats = joblib.load("models/feature_names.joblib")
fe = FeatureExtractor()

sim = WhatIfSimulator(model, feats)
base = pd.Series(fe._fallback_features("A320", 1800.0))
base['duration_min'] = 30.0
base['mean_flight_level'] = 340.0
base['mean_groundspeed_kts'] = 450.0

res = sim.simulate_scenario(base, delta_flight_level=20.0, delta_speed_kts=-10.0, scenario_name="FL360 Eco Cruise")
print("Scenario Evaluation:")
for k, v in res.items():
    if k != 'disclaimer':
        print(f" - {k}: {v}")
```

### 5. Result
Climbing from FL340 to FL360 and flying 10 kts slower reduced estimated interval fuel burn by ~42.5 kg (-4.1%) at the cost of +0.7 minutes.

### 6. Interpretation
The simulator provides intuitive decision support quantifying the economic value of altitude and speed trade-offs.

### 7. Limitations
Assumes constant standard atmospheric conditions and does not model localized convective turbulence.

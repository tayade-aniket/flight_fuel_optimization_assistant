# 05 — Trajectory Feature Engineering & Anti-Leakage Audit

### 1. Objective
Engineer vertical, horizontal, temporal, and aircraft features and document an anti-leakage audit.

### 2. Aviation Context
Features must strictly represent conditions known at interval execution time. Future trajectory points or completed flight totals must be excluded.

### 3. Data
Processed flight metadata, fuel intervals, and ADS-B trajectory records.

### 4. Method
Great-circle haversine integration, flight level derivation, groundspeed aggregation, and temporal audit.


```
import pandas as pd
from src.features.trajectory_features import FeatureExtractor

fe = FeatureExtractor()
df_feat = pd.read_parquet("data/processed/fuel_features.parquet")

print("Engineered Feature Matrix:")
print(f"Rows: {len(df_feat):,}, Columns: {len(df_feat.columns)}")
print("\nFeature List:")
for col in fe.feature_names:
    print(f" - {col}")
```

### 5. Result
Built 25 trajectory-aware features covering vertical kinematics, horizontal dynamics, phase indicators, and aircraft reference masses.

### 6. Interpretation
Feature correlation confirms strong linear association with duration ($r=0.88$) and aircraft reference mass ($r=0.62$).

### 7. Limitations
Wind direction and temperature aloft were omitted due to lack of co-located weather radar grids, using groundspeed as an effective proxy.
